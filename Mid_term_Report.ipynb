$$ SSIM(x,y)=\frac{(2\mu_x \mu_y + C_1) * (2\sigma_{xy} + C_2)} {(\mu_x^2 + \mu_y^2 + C_1) * (\sigma_x^2 + \sigma_y^2 + C_2)}$$


where
*   $\mu_x$ and $\mu_y$ are means of images $x$ and $y$ respectively
*   $\sigma_x$ and $\sigma_y$ are the standard deviations of images $x$ and $y$ respectively
*   $\sigma_{xy}$ is the covariance of images $x$ and $y$
*   $C_1$ and $C_2$ are small constants to stabilize the division with weak denominator.

**The SSIM value ranges from -1 to 1, where 1 indicates perfect similarity between the images.**

## Question 2: Find correlation between two images with Pearson Correlation Coefficient (20 points)

Complete the following program to compute pearson correlation coefficient between two images by using the following images.

pearson correlation coefficient equation is as follows

$$
r = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}
         {\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2} \cdot \sqrt{\sum_{i=1}^{n}(y_i - \bar{y})^2}}
$$

or vector notation:

$$
r = \frac{\text{cov}(x, y)}{\sigma_x \sigma_y}
$$

where


*   $x$ and $y$ are matrices or vectors
* $\bar{x}$ and $\bar{y}$ are the means of $x$ and $y$  
* $\sigma_x$ and $\sigma_y$ are the standard deviations of $x$ and $y$  
* $\text{cov}(x, y)$ is the covariance between $x$ and $y$
